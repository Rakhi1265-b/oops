````python
class BankAccount:
    def __init__(self,account_holder,balance):
        self.account_holder=account_holder
        self.balance=balance
    def deposit(self,amount):
        self.balance+=amount
    
    def withdraw(self,amount):
        self.balance-=amount
    def __str__(self):
        return str(self.account_holder)
    def __add__(self, other):
        return (self.balance+other.balance)
    def __sub__(self, other):
        return (self.balance-other.balance)
    def __eq__(self, value):
        return (self.balance==value.balance)
    def __lt__(self, other):
        return (self.balance<other.balance)

a=BankAccount("Rakesh",50000)
b=BankAccount("Raistar",30000)
a.deposit(1000)
b.deposit(2000)
a.withdraw(1000)
b.deposit(2000)
add=a+b
print(add)
sub=a-b
print(sub)
print(a==b)
print(a<b)    
````

# Question 2: Product Price Comparison

Create a class `Product` with:

- Attributes:
  - `name`
  - `price`
  - `quantity`

- Method:
  - `total_price()`

## Implement the following magic methods:

- `__str__()` → Display the product details.
- `__add__()` → Add the total prices of two products.
- `__mul__()` → Multiply the product price by a number.
- `__gt__()` → Compare which product has a greater total value.
- `__eq__()` → Compare the prices of two products.
- `__getattr__()` → Return `"Attribute not found"` for missing attributes.
- `__setattr__()` → Do not allow the price to be less than `0`.

## Solution

````python
class Product:
    def __init__(self,name,price,quantity):
        self.name=name
        self.price=price
        self.quantity=quantity

    def total(self):
        return self.price*self.quantity
    def __str__(self):
        return f"{self.name}\n {self.total}\n {self.quantity}"
    def __add__(self, other):
        return (self.total()+other.total())
    def __mul__(self,number):
        return (self.total()*number)
    def __gt__(self, other):
        return (self.total() > other.total())
    def __eq__(self, value):
        return (self.total==value.total)
a=Product("Boost",5,10)
b=Product("Complain",10,3)        
add=a+b
print(add)
mul=a*2
print(mul)
print(a>b)
print(a==b)
````

````python
from abc import ABC, abstractmethod


class Account(ABC):

    # Bank-wide interest policies
    savings_rate = 5
    current_rate = 2
    fixed_deposit_rate = 7

    def __init__(self, account_number, holder_name, balance):
        self.__account_number = account_number
        self.__holder_name = holder_name
        self.__balance = balance

    # Account number property
    @property
    def account_number(self):
        return self.__account_number

    # Holder name property
    @property
    def holder_name(self):
        return self.__holder_name

    # Balance property
    @property
    def balance(self):
        return self.__balance

    # Deposit
    def deposit(self, amount):
        if Account.validate_amount(amount):
            self.__balance += amount
            print("Amount deposited:", amount)
        else:
            print("Invalid amount")

    # Withdraw
    def withdraw(self, amount):
        if not Account.validate_amount(amount):
            print("Invalid amount")
        elif amount > self.__balance:
            print("Insufficient balance")
        else:
            self.__balance -= amount
            print("Amount withdrawn:", amount)

    # Static method
    @staticmethod
    def validate_amount(amount):
        return amount > 0

    # Class method
    @classmethod
    def update_interest_policy(cls, savings, current, fixed_deposit):
        cls.savings_rate = savings
        cls.current_rate = current
        cls.fixed_deposit_rate = fixed_deposit

        print("Interest rates updated")

    # Abstract method
    @abstractmethod
    def calculate_interest(self):
        pass

    # Display details
    def display_details(self):
        print("Account Number :", self.account_number)
        print("Holder Name    :", self.holder_name)
        print("Balance        :", self.balance)
        print("Interest       :", self.calculate_interest())


# Savings Account
class SavingsAccount(Account):

    def calculate_interest(self):
        return self.balance * Account.savings_rate / 100


# Current Account
class CurrentAccount(Account):

    def calculate_interest(self):
        return self.balance * Account.current_rate / 100


# Fixed Deposit Account
class FixedDepositAccount(Account):

    def calculate_interest(self):
        return self.balance * Account.fixed_deposit_rate / 100


# Creating objects
s = SavingsAccount(101, "Rakesh", 20000)

c = CurrentAccount(102, "Rakhi", 30000)

f = FixedDepositAccount(103, "Hari", 50000)


# Deposit
s.deposit(5000)

# Withdraw
s.withdraw(2000)

print()

# Display details
s.display_details()

print()

c.display_details()



f.display_details()
````
