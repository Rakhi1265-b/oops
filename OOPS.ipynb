```python
class Student:
    def __init__(self,name,id,branch):
        self.name=name
        self.id=id
        self.branch=branch
    def display(self):
        print(self.name)
        print(self.id)
        print(self.branch)
class profiling(Student):
    def profile(self):
        print("DETAILS OF STUDENTS")
    def __init__(self, name, id, branch,percentage):
        super().__init__(name, id, branch)
        self.percentage=percentage
    def display(self):
        super().display()
        print(self.percentage)
        

s1=profiling("Rakesh","B201265","ECE",69.01)
s1.profile()
s1.display()
s2=Student("Raistar","B201222","CSE")
print()
s2.display()

```

```python
class Cse:
    def __init__(self):
        self.cutoff=9.80
        self.percentt="98.00%"
        
    def about(self):
        print("COMPUTER SCIENCE AND ENGINEERING")
    def oppertunity(self):
        print("Testing \nDeveloper \nR&D")
    
class Ece:
    def __init__(self):
        self.cutof=9.51
        self.percent="95.10%"
        
    def detial(self):
        print("ELECTRIONICS AND COMMUNICATIONS ENGINEERING ")
    def oppertunity(self):
        print("Designing \nWireless communications \nR&D")
class Eng(Cse,Ece):
      def __init__(self,category):
          Cse.__init__(self)
          Ece.__init__(self)
          self.category=category
          
      def display(self):
          print("The Category you are:",self.category)
      def qualify(self, cgpa):

        if cgpa >= 9.8:
            print("congrats! your selected for CSE and ECE. ")
            print("\n--- CSE ---")
            self.about()
            print("Cutoff:", self.cutoff)
            print("Percentage:", self.percentt)
            print("OPPORTUNITIES:")
            Cse.oppertunity(self)

            print("\n--- ECE ---")
            self.detial()
            print("Cutoff:", self.cutof)
            print("Percentage:", self.percent)
            print("OPPORTUNITIES:")
            Ece.oppertunity(self)

        elif cgpa >= 9.6:
            print("congrats! your selected for ECE. ")

            print("\n--- ECE ---")
            self.detial()
            print("Cutoff:", self.cutof)
            print("Percentage:", self.percent)
            print("OPPORTUNITIES:")
            Ece.oppertunity(self)

        else:

            print("You are not qualified for any branch")

your_cgpa=float(input("Enter your cgpa: "))
clg=input("Enter prefered University: ")
p=Eng(clg)
print("the University your selected :",clg)
p.qualify(your_cgpa)
```
