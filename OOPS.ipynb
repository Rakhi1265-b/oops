```python
# Questions On Generator
# 1.	Write a generator that yields numbers from 1 to N.
def u(start,end):
    while(start<=end):
        value=start
        yield value
        start+=1
for i in u(1,10):
    print(i)

```

```python
# 3.	Write a generator that yields each character of a string.
def gen(st):
    start=st
    index=0
    while(index<len(start)):
        yield start[index]
        index+=1
for i in gen(list("rakesh")):
    print(i)
```

```python

# 8.	Write a generator that yields digits from an integer one by one.
def gen(st):
    start=st
    m=0
    while(start!=0):
        r=start%10
        m=m*10+r
        start=start//10
    start=m
    while(start!=0):
        s=start%10
        yield s
        start=start//10
l=143
for i in gen(l):
    print(i,end=" ")
```

```python

# 9.	Create a generator that yields cumulative sum of numbers in a list. Example: [1,2,3] → 1, 3, 6
def gen(start):
    sum=0
    i=0
    while(i<len(start)):
        sum=sum+start[i]
        yield sum
        i+=1
for i in gen([1,2,3]):
    print(i,end=" ")
```

```python
# Questions On Iterator

# 1.	Create an custom iterator that prints numbers from 1 to N, where N is given by the user.

class counter:
    def __init__(self,end,st):
        self.end=end
        self.st=st
    def __iter__(self):
        return self
    def __next__(self):
        while(self.st<self.end):
            value=self.st
            self.st+=1
            return value
        raise StopIteration
for i in counter(10,1):
    print(i)
            
```
